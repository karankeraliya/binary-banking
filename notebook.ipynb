```python
# Fix joblib pickling by defining droppers as top-level callables with closure-free params.

import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.model_selection import train_test_split
import joblib
import textwrap

df = pd.read_csv("train.csv")
y = df["y"]
X = df.drop(columns=["y"])

def add_pdays_features(X_df: pd.DataFrame) -> pd.DataFrame:
    X_df = X_df.copy()
    X_df["was_contacted_before"] = (X_df["pdays"] != -1).astype(int)
    X_df["pdays_clean"] = X_df["pdays"].replace(-1, np.nan)
    median_pdays = X_df["pdays_clean"].median(skipna=True)
    X_df["pdays_clean"] = X_df["pdays_clean"].fillna(median_pdays)
    return X_df

# Define two explicit droppers to avoid closures (pickleable)
def drop_cols_with_duration(X_df: pd.DataFrame) -> pd.DataFrame:
    return X_df.drop(columns=[c for c in ["id", "pdays"] if c in X_df.columns])

def drop_cols_without_duration(X_df: pd.DataFrame) -> pd.DataFrame:
    return X_df.drop(columns=[c for c in ["id", "pdays", "duration"] if c in X_df.columns])

def build_preprocessor(include_duration: bool = True) -> Pipeline:
    pre_steps = [("add_pdays", FunctionTransformer(add_pdays_features, feature_names_out="pandas"))]

    if include_duration:
        pre_steps.append(("drop_cols", FunctionTransformer(drop_cols_with_duration, feature_names_out="pandas")))
        probe = drop_cols_with_duration(add_pdays_features(X.copy()))
    else:
        pre_steps.append(("drop_cols", FunctionTransformer(drop_cols_without_duration, feature_names_out="pandas")))
        probe = drop_cols_without_duration(add_pdays_features(X.copy()))

    cat_after = probe.select_dtypes(include=["object"]).columns.tolist()
    num_after = probe.select_dtypes(include=["int64", "float64"]).columns.tolist()

    col_tf = ColumnTransformer(
        transformers=[
            ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_after),
            ("num", StandardScaler(), num_after),
        ],
        remainder="drop",
        verbose_feature_names_out=False,
    )
    pipe = Pipeline(steps=pre_steps + [("encode_scale", col_tf)])
    return pipe

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

pre_with_duration = build_preprocessor(include_duration=True)
pre_without_duration = build_preprocessor(include_duration=False)

# Test fit-transform on a modest sample
sample_idx = np.random.RandomState(42).choice(len(X_train), size=5000, replace=False)
Xt_sample = pre_with_duration.fit_transform(X_train.iloc[sample_idx], y_train.iloc[sample_idx])

# Save artifacts (now picklable)
joblib.dump(pre_with_duration, "/mnt/data/preprocessor_with_duration.joblib")
joblib.dump(pre_without_duration, "/mnt/data/preprocessor_without_duration.joblib")

# Save README
readme = textwrap.dedent("""
Reusable scikit-learn preprocessing pipeline for the Kaggle Bank Marketing-style dataset
=======================================================================================

Files saved:
- preprocessor_with_duration.joblib
- preprocessor_without_duration.joblib

How to use:
-----------
from sklearn.linear_model import LogisticRegression
import joblib
import pandas as pd

df = pd.read_csv("train.csv")
X = df.drop(columns=["y"])
y = df["y"]

pre = joblib.load("preprocessor_with_duration.joblib")  # or preprocessor_without_duration.joblib
X_proc = pre.fit_transform(X, y)  # or pre.transform(X) if already fit

from sklearn.linear_model import LogisticRegression
clf = LogisticRegression(max_iter=1000, class_weight="balanced", n_jobs=-1)
clf.fit(X_proc, y)
""").strip()
with open("README_preprocessing.txt", "w") as f:
    f.write(readme)

Xt_sample.shape
```

```python
# preprocessing.py
# ------------------------------------------------------------
# Production-style preprocessing for the Bank Marketing dataset
# - No probing / no duplicate function calls
# - Clear, explicit column lists
# - Pickle-friendly custom transformers
# - Works with sklearn <1.2 and >=1.2 (no sparse_output arg)
# ------------------------------------------------------------

from dataclasses import dataclass
from typing import List, Optional

import numpy as np
import pandas as pd

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn import set_config


# --- Columns (explicit & easy to reason about) -------------------------------

CATEGORICAL_COLS: List[str] = [
    "job", "marital", "education", "default",
    "housing", "loan", "contact", "month", "poutcome"
]

# Raw numeric cols in the CSV
RAW_NUMERIC_COLS: List[str] = [
    "age", "balance", "day", "duration", "campaign", "pdays", "previous"
]

ID_COL = "id"
TARGET_COL = "y"


# --- Custom transformers (pickle-friendly) -----------------------------------

class AddPdaysFeatures(BaseEstimator, TransformerMixin):
    """
    Adds:
      - was_contacted_before = (pdays != -1) as int
      - pdays_clean: replaces -1 with NaN, then median-imputes
    Leaves all original columns intact (we'll drop 'pdays' later).
    """
    def fit(self, X: pd.DataFrame, y: Optional[pd.Series] = None):
        return self

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        X = X.copy()
        X["was_contacted_before"] = (X["pdays"] != -1).astype(int)
        pdays_clean = X["pdays"].replace(-1, np.nan)
        X["pdays_clean"] = pdays_clean.fillna(pdays_clean.median(skipna=True))
        return X


@dataclass
class DropColumns(BaseEstimator, TransformerMixin):
    """
    Drops: id, pdays, and optionally duration (to avoid leakage in real-time use).
    """
    include_duration: bool = True

    def fit(self, X: pd.DataFrame, y: Optional[pd.Series] = None):
        return self

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        X = X.copy()
        cols_to_drop = [ID_COL, "pdays"]
        if not self.include_duration:
            cols_to_drop.append("duration")
        cols_to_drop = [c for c in cols_to_drop if c in X.columns]
        return X.drop(columns=cols_to_drop)


# --- Builder -----------------------------------------------------------------

def build_preprocessor(include_duration: bool = True) -> Pipeline:
    """
    Returns a sklearn Pipeline that:
      1) adds pdays features,
      2) drops id/pdays/(duration),
      3) one-hot encodes categoricals,
      4) standard-scales numerics.
    """
    # After AddPdaysFeatures + DropColumns, these will exist:
    final_numeric_cols = [
        "age", "balance", "day",
        *(["duration"] if include_duration else []),
        "campaign", "previous",
        "was_contacted_before", "pdays_clean"
    ]
    final_categorical_cols = CATEGORICAL_COLS[:]  # unchanged

    # Encoders/scalers (keep dense output to be simple & compatible)
    ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    scaler = StandardScaler()

    col_tf = ColumnTransformer(
        transformers=[
            ("cat", ohe, final_categorical_cols),
            ("num", scaler, final_numeric_cols),
        ],
        remainder="drop",
        verbose_feature_names_out=False,
    )

    pipe = Pipeline(steps=[
        ("add_pdays", AddPdaysFeatures()),
        ("drop_cols", DropColumns(include_duration=include_duration)),
        ("encode_scale", col_tf),
    ])

    # Make all transformers pass pandas DataFrames between steps so column names line up
    set_config(transform_output="pandas")
    # Removed: pipe.set_output(transform="pandas")  # This caused the error
    return pipe


# --- Example usage -----------------------------------------------------------
if __name__ == "__main__":
    import joblib
    from sklearn.model_selection import train_test_split

    df = pd.read_csv("train.csv")
    y = df[TARGET_COL]
    X = df.drop(columns=[TARGET_COL])

    X_train, X_val, y_train, y_val = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )

    # Build both variants
    pre_with_duration = build_preprocessor(include_duration=True)
    pre_without_duration = build_preprocessor(include_duration=False)

    # Fit-transform a small sample just to validate end-to-end quickly
    idx = np.random.RandomState(42).choice(len(X_train), size=5000, replace=False)
    Xt_sample = pre_with_duration.fit_transform(X_train.iloc[idx], y_train.iloc[idx])
    print("Sample transformed shape (with duration):", Xt_sample.shape)

    # Save
    joblib.dump(pre_with_duration, "preprocessor_with_duration.joblib")
    joblib.dump(pre_without_duration, "preprocessor_without_duration.joblib")
```
